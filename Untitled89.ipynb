```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sqlalchemy import create_engine

# Load data
df = pd.read_csv("customer_shopping_behavior.csv")

# Data overview
df.info()
df.head()
df.describe()

# Handle missing values
df["Review Rating"] = df.groupby("Category")["Review Rating"].transform(lambda x: x.median())

# Clean column names
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')

# Create age groups
label = ["young", "adult", "senior", "aged"]
df["age_group"] = pd.qcut(df["age"], q=4, labels=label)

# Create frequency mapping
frequency = {
    "Fortnightly": 14,
    "Weekly": 7,
    "Bi-Weekly": 14,
    "Monthly": 30,
    "Quarterly": 90,
    "Every 3 Months": 90,
    "Annually": 365
}
df["frequency"] = df["frequency_of_purchases"].map(frequency)
df['purchase_amount'] = df['purchase_amount_(usd)']
df.drop("purchase_amount_(usd)", inplace=True, axis=1)

# ============ VISUALIZATIONS ============

# Set style
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")

# 1. Q1: Revenue by Gender
plt.figure(figsize=(10, 6))
gender_revenue = df.groupby('gender')['purchase_amount'].sum().sort_values(ascending=False)
bars = plt.bar(gender_revenue.index, gender_revenue.values, color=['#FF6B6B', '#4ECDC4'])
plt.title('Total Revenue by Gender', fontsize=16, fontweight='bold')
plt.xlabel('Gender', fontsize=12)
plt.ylabel('Total Revenue (USD)', fontsize=12)
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height,
             f'${height:,.0f}', ha='center', va='bottom', fontsize=11)
plt.tight_layout()
plt.savefig('1_gender_revenue.png', dpi=300, bbox_inches='tight')
plt.show()

# 2. Q5: Subscribers vs Non-subscribers
plt.figure(figsize=(12, 5))

# Subplot 1: Average Spend
plt.subplot(1, 2, 1)
sub_avg = df.groupby('subscription_status')['purchase_amount'].mean().sort_values(ascending=False)
bars = plt.bar(sub_avg.index, sub_avg.values, color=['#FFA94D', '#74C0FC'])
plt.title('Average Spend by Subscription Status', fontsize=14, fontweight='bold')
plt.xlabel('Subscription Status', fontsize=11)
plt.ylabel('Average Purchase Amount (USD)', fontsize=11)
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height,
             f'${height:.2f}', ha='center', va='bottom', fontsize=10)

# Subplot 2: Total Revenue
plt.subplot(1, 2, 2)
sub_total = df.groupby('subscription_status')['purchase_amount'].sum().sort_values(ascending=False)
bars = plt.bar(sub_total.index, sub_total.values, color=['#FFA94D', '#74C0FC'])
plt.title('Total Revenue by Subscription Status', fontsize=14, fontweight='bold')
plt.xlabel('Subscription Status', fontsize=11)
plt.ylabel('Total Revenue (USD)', fontsize=11)
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height,
             f'${height:,.0f}', ha='center', va='bottom', fontsize=10)
plt.tight_layout()
plt.savefig('2_subscription_impact.png', dpi=300, bbox_inches='tight')
plt.show()

# 3. Q3: Top 5 Products by Review Rating
plt.figure(figsize=(12, 6))
top_products = df.groupby('item_purchased')['review_rating'].mean().sort_values(ascending=False).head(5)
colors = plt.cm.viridis(np.linspace(0.3, 0.9, 5))
bars = plt.barh(top_products.index[::-1], top_products.values[::-1], color=colors[::-1])
plt.title('Top 5 Products with Highest Average Review Rating', fontsize=16, fontweight='bold')
plt.xlabel('Average Review Rating', fontsize=12)
plt.ylabel('Product', fontsize=12)
for bar in bars:
    width = bar.get_width()
    plt.text(width + 0.02, bar.get_y() + bar.get_height()/2.,
             f'{width:.2f}', ha='left', va='center', fontsize=11)
plt.xlim(0, 5.5)
plt.tight_layout()
plt.savefig('3_top_products_ratings.png', dpi=300, bbox_inches='tight')
plt.show()

# 4. Q7: Customer Segmentation
plt.figure(figsize=(10, 6))
df_segment = df.copy()
df_segment['customer_segment'] = pd.cut(df_segment['previous_purchases'],
                                        bins=[0, 1, 10, float('inf')],
                                        labels=['New', 'Returning', 'Loyal'])
segment_counts = df_segment['customer_segment'].value_counts()
colors = ['#FF6B6B', '#FECA57', '#48DBFB']
explode = (0.05, 0.05, 0.1)
plt.pie(segment_counts.values, labels=segment_counts.index, autopct='%1.1f%%',
        colors=colors, explode=explode, shadow=True, startangle=90,
        textprops={'fontsize': 12, 'fontweight': 'bold'})
plt.title('Customer Segmentation Distribution', fontsize=16, fontweight='bold')
plt.legend([f'{label}: {count:,} customers' for label, count in segment_counts.items()],
           loc='upper right', fontsize=10)
plt.tight_layout()
plt.savefig('4_customer_segmentation.png', dpi=300, bbox_inches='tight')
plt.show()

# 5. Q10: Revenue by Age Group
plt.figure(figsize=(10, 6))
age_revenue = df.groupby('age_group')['purchase_amount'].sum().sort_values(ascending=False)
age_counts = df.groupby('age_group').size()
colors = plt.cm.Reds(np.linspace(0.4, 0.9, 4))
bars = plt.bar(age_revenue.index, age_revenue.values, color=colors)
plt.title('Revenue Contribution by Age Group', fontsize=16, fontweight='bold')
plt.xlabel('Age Group', fontsize=12)
plt.ylabel('Total Revenue (USD)', fontsize=12)
for bar, age in zip(bars, age_revenue.index):
    height = bar.get_height()
    revenue = age_revenue[age]
    count = age_counts[age]
    plt.text(bar.get_x() + bar.get_width()/2., height + 500,
             f'${revenue:,.0f}\n(n={count:,})', ha='center', va='bottom', fontsize=10)
plt.tight_layout()
plt.savefig('5_age_group_revenue.png', dpi=300, bbox_inches='tight')
plt.show()

# 6. Q6: Products with Highest Discount Rate
plt.figure(figsize=(12, 6))
discount_rate = df.groupby('item_purchased').apply(
    lambda x: 100 * (x['discount_applied'] == 'Yes').sum() / len(x)
).sort_values(ascending=False).head(5)
colors = plt.cm.Greens(np.linspace(0.3, 0.8, 5))
bars = plt.barh(discount_rate.index[::-1], discount_rate.values[::-1], color=colors[::-1])
plt.title('Top 5 Products with Highest Discount Application Rate', fontsize=16, fontweight='bold')
plt.xlabel('Discount Rate (%)', fontsize=12)
plt.ylabel('Product', fontsize=12)
for bar in bars:
    width = bar.get_width()
    plt.text(width + 0.5, bar.get_y() + bar.get_height()/2.,
             f'{width:.1f}%', ha='left', va='center', fontsize=11)
plt.tight_layout()
plt.savefig('6_discount_rate_products.png', dpi=300, bbox_inches='tight')
plt.show()

# 7. Q8: Top 3 Products per Category
plt.figure(figsize=(14, 8))
categories = df['category'].unique()
colors = plt.cm.Set3(np.linspace(0, 1, len(categories)))
for i, category in enumerate(categories):
    top_products_cat = df[df['category'] == category]['item_purchased'].value_counts().head(3)
    plt.subplot(2, 3, i+1)
    bars = plt.bar(top_products_cat.index, top_products_cat.values, color=colors[i])
    plt.title(f'Top 3 in {category}', fontsize=12, fontweight='bold')
    plt.xlabel('Product', fontsize=9)
    plt.ylabel('Orders Count', fontsize=9)
    plt.xticks(rotation=45, ha='right', fontsize=8)
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2., height,
                 f'{int(height)}', ha='center', va='bottom', fontsize=8)
plt.tight_layout()
plt.savefig('7_top_products_per_category.png', dpi=300, bbox_inches='tight')
plt.show()

# 8. Q4: Shipping Type Comparison
plt.figure(figsize=(8, 6))
shipping_avg = df[df['shipping_type'].isin(['Standard', 'Express'])].groupby('shipping_type')['purchase_amount'].mean()
colors = ['#FF9FF3', '#54A0FF']
bars = plt.bar(shipping_avg.index, shipping_avg.values, color=colors)
plt.title('Average Purchase Amount by Shipping Type', fontsize=14, fontweight='bold')
plt.xlabel('Shipping Type', fontsize=12)
plt.ylabel('Average Purchase Amount (USD)', fontsize=12)
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height + 0.5,
             f'${height:.2f}', ha='center', va='bottom', fontsize=11)
plt.tight_layout()
plt.savefig('8_shipping_comparison.png', dpi=300, bbox_inches='tight')
plt.show()

# 9. Q9: Repeat Buyers Subscription Pattern
plt.figure(figsize=(8, 6))
repeat_buyers = df[df['previous_purchases'] > 5].groupby('subscription_status').size()
colors = ['#F368E0', '#00D2D3']
bars = plt.bar(repeat_buyers.index, repeat_buyers.values, color=colors)
plt.title('Subscription Status of Repeat Buyers (>5 Purchases)', fontsize=14, fontweight='bold')
plt.xlabel('Subscription Status', fontsize=12)
plt.ylabel('Number of Repeat Buyers', fontsize=12)
for bar in bars:
    height = bar.get_height()
    percentage = (height / repeat_buyers.sum()) * 100
    plt.text(bar.get_x() + bar.get_width()/2., height + 1,
             f'{int(height)}\n({percentage:.1f}%)', ha='center', va='bottom', fontsize=10)
plt.tight_layout()
plt.savefig('9_repeat_buyers_subscription.png', dpi=300, bbox_inches='tight')
plt.show()

# 10. Overall Revenue Distribution by Category
plt.figure(figsize=(12, 6))
category_revenue = df.groupby('category')['purchase_amount'].sum().sort_values(ascending=False)
colors = plt.cm.Paired(np.linspace(0, 1, len(category_revenue)))
bars = plt.bar(category_revenue.index, category_revenue.values, color=colors)
plt.title('Revenue Distribution by Product Category', fontsize=16, fontweight='bold')
plt.xlabel('Category', fontsize=12)
plt.ylabel('Total Revenue (USD)', fontsize=12)
for bar in bars:
    height = bar.get_height()
    percentage = (height / category_revenue.sum()) * 100
    plt.text(bar.get_x() + bar.get_width()/2., height + 500,
             f'${height:,.0f}\n({percentage:.1f}%)', ha='center', va='bottom', fontsize=10)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('10_category_revenue.png', dpi=300, bbox_inches='tight')
plt.show()

# 11. Heatmap: Age vs Purchase Amount
plt.figure(figsize=(12, 8))
age_pivot = df.pivot_table(values='purchase_amount', index='gender', columns='age_group', aggfunc='mean')
sns.heatmap(age_pivot, annot=True, fmt='.0f', cmap='YlOrRd', linewidths=0.5,
            cbar_kws={'label': 'Average Purchase Amount (USD)'})
plt.title('Average Purchase Amount by Gender and Age Group', fontsize=16, fontweight='bold')
plt.xlabel('Age Group', fontsize=12)
plt.ylabel('Gender', fontsize=12)
plt.tight_layout()
plt.savefig('11_heatmap_gender_age.png', dpi=300, bbox_inches='tight')
plt.show()

# 12. Correlation Matrix
plt.figure(figsize=(10, 8))
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns
corr_matrix = df[numeric_cols].corr()
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
sns.heatmap(corr_matrix, mask=mask, annot=True, fmt='.2f', cmap='coolwarm',
            square=True, linewidths=0.5, cbar_kws={'label': 'Correlation'})
plt.title('Correlation Matrix of Numerical Features', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.savefig('12_correlation_matrix.png', dpi=300, bbox_inches='tight')
plt.show()

# Export to PostgreSQL
from sqlalchemy import create_engine

username = 'postgres'
password = 'anjal'
host = 'localhost'
database = "customer_behavior"

engine = create_engine(
    f"postgresql://{username}:{password}@{host}:5432/{database}"
)
table_name = 'customer'
df.to_sql(table_name, engine, if_exists='replace', index=False)
```
