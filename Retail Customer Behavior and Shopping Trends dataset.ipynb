```python
df["Review Rating"] = df.groupby("Category")["Review Rating"].transform(lambda x: x.median())
```

```python
df.isnull().sum()


```

```python
df.head()
```

```python
df.columns = df.columns.str.lower()
```

```python
df.columns = df.columns.str.replace(' ', '_')
```

```python
df.columns
```

```python
label = ["young","adult","senior","aged"]
df["age_group"]= pd.qcut(df["age"],q=4 , labels = label )
```

```python
df.head()
```

```python
df['frequency_of_purchases'].unique()
```

```python
frequency = {
    "Fortnightly": 14,
    "Weekly": 7,
    "Bi-Weekly": 14,
    "Monthly": 30,
    "Quarterly": 90,
    "Every 3 Months": 90,
    "Annually": 365
}

df["frequency"] = df["frequency_of_purchases"].map(frequency)
```

```python
df.head()
```

```python
df['purchase_amount'] = df['purchase_amount_(usd)']
```

```python
df.drop("purchase_amount_(usd)", inplace =True, axis = 1)
```
